## Apply Count Vectorizer to Our Spam, Ham Example

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
spamham_countvectorizer = vectorizer.fit_transform(spam_df['text'])
```

```python
print(vectorizer.get_feature_names_out())
```

```python
print(spamham_countvectorizer.toarray())
```

```python
spamham_countvectorizer.shape
```

# Step 5: Training The Model

```python
label = spam_df['spam'].values
```

```python
label
```

```python
from sklearn.naive_bayes import MultinomialNB

NB_Classifier = MultinomialNB()
NB_Classifier.fit(spamham_countvectorizer, label)
```

```python
testing_sample = ['Free money!!', 'Hi Kim, Please let me know if you need any further information. Thanks']
testing_sample_countvectorizer = vectorizer.transform(testing_sample)
```

```python
test_predict = NB_Classifier.predict(testing_sample_countvectorizer)
test_predict
```

```python
testing_sample = ['Hello, I am Ryan, i would like to book a hotel in Bali by January 24th', 'money viagra!!!!!!']
testing_sample_countvectorizer = vectorizer.transform(testing_sample)
test_predict = NB_Classifier.predict(testing_sample_countvectorizer)
test_predict
```

# Step 5: Divide The Data Into Training and Testing Prop to Training

```python
X = spamham_countvectorizer
y = label
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
```

```python
from sklearn.naive_bayes import MultinomialNB

NB_Classifier = MultinomialNB()
NB_Classifier.fit(X_train, y_train)
```

# Step 6: Evaluation The Model

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
y_predict_train = NB_Classifier.predict(X_train)
y_predict_train
```

```python
cm = confusion_matrix(y_train, y_predict_train)
sns.heatmap(cm, annot = True)
```

```python
y_predict_test = NB_Classifier.predict(X_test)
cm = confusion_matrix(y_test, y_predict_test)
sns.heatmap(cm, annot= True)
```

```python
print(classification_report(y_test, y_predict_test))
```
